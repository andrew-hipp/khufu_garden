```r
library(openxlsx)
library(stringr)

dat.kh <- read.xlsx('../data/OakSpmExtract-KhufuTips-20240513.xlsx')
dat.spm <- read.xlsx('../data/AAA.Quercus_Specimen_Database_LIVE.xlsx')
dat.ext <- read.xlsx('../data/Syst.DNA.database.LIVE.xlsx')
dat.submitted <- read.xlsx('../data/khufu_submitted_AHcleaned.xlsx')

headers.missing <- names(dat.kh)[which(!names(dat.kh) %in% names(dat.spm))]

samples.submitted <- dat.submitted$dat.submitted$sampleID
samples.submitted <- gsub('-', '.', dat.submitted$sampleID, fixed= T)

index = grep('OAK|SYST', samples.submitted, invert = T)
samples.submitted[index] <- 
    paste('MOR',
          str_pad(gsub('MOR', '', samples.submitted[index]), 4, pad = "0"),
          sep = ''
          )

samples.add <- setdiff(gsub('-', '.', samples.submitted, fixed= T), dat.kh$khufuTips) |>
    grep(patt = '.', fixed = T, value = T)
samples.add.spm <- dat.ext$'PROJECT-Specimen_ID_CODE'[match(gsub('.', '-', samples.add, fixed = T), dat.ext$MOR_DNA_TUBE_NUMBER)]
new.table <- dat.spm[match(samples.add.spm, dat.spm$SPMCODE), setdiff(names(dat.kh), headers.missing)]
# headers.missing
new.table$MOTHER <- NA
new.table$Site_georef <- new.table$Site
new.table$'TEMP-Khufu' <- samples.add.spm
new.table$khufuTips <- samples.add
new.table <- new.table[names(dat.kh)]

new.table <- rbind(dat.kh, new.table)
write.csv(new.table, '../data/OakSpmExtract-KhufuTips-20241114.csv', row.names = F)
names(new.table)
```
