### Convert vcf files to bed and phylip
Both vcf files were converted to bed files using `plink`. Both are thinned to one snp per 5000 bp (LD tails off at 500 bp in _Quercus_; Plomion et al., 2016 -- AH check this reference). We also convert to relaxed Phylip format to get genetic distances using RAxML. We are going to filter these in a few ways:

* Exclude two samples of _Quercus gambelii_ from both files, because those were included only as an outgroup for tree-rooting. The gambelii individuals are designated in `data/gambelii.txt`.
* Generate two sets of imputed data, one with all individuals, one pruning out individuals for which < 10% of SNPs were called

First, the imputed data, all individuals:

```bash
conda activate garden
```

```bash
FILE=KHU075_SNPcalls_20241105.imputed

plink --vcf ../data/KHUFU/$FILE.vcf.gz --make-bed --remove ../data/gambelii.txt --out ../data/bed/$FILE --allow-extra-chr
plink --bfile ../data/bed/$FILE --bp-space 5000 --make-bed --out ../data/bed/$FILE.bp5000 --allow-extra-chr
plink2 --bfile ../data/bed/$FILE --bp-space 5000 --export phylip --out ../data/bed/$FILE.bp5000 --allow-extra-chr

awk '{$1="0";print $0}' ../data/bed/$FILE.bp5000.bim > ../data/bed/$FILE.bp5000.bim.tmp
mv ../data/bed/$FILE.bp5000.bim.tmp ../data/bed/$FILE.bp5000.bim
```

... and then imputed data without the individuals with < 10% of SNPs imputed:

```bash
FILE=KHU075_SNPcalls_20241105.imputed

plink --vcf ../data/KHUFU/$FILE.vcf.gz --make-bed --remove ../data/excludeLowSnps.txt --out ../data/bed/$FILE.m90 --allow-extra-chr
plink --bfile ../data/bed/$FILE.m90 --bp-space 5000 --make-bed --out ../data/bed/$FILE.m90.bp5000 --allow-extra-chr
plink2 --bfile ../data/bed/$FILE.m90 --bp-space 5000 --export phylip --out ../data/bed/$FILE.m90.bp5000 --allow-extra-chr

awk '{$1="0";print $0}' ../data/bed/$FILE.m90.bp5000.bim > ../data/bed/$FILE.m90.bp5000.bim.tmp
mv ../data/bed/$FILE.m90.bp5000.bim.tmp ../data/bed/$FILE.m90.bp5000.bim
```

and then the raw data. For this, we also eliminate first columns then individuals that are missing more than 90% of data.

```bash
FILE=KHU075_SNPcalls_20241105.raw

plink --vcf ../data/KHUFU/$FILE.vcf.gz --make-bed --geno 0.9 --mind 0.9 --remove ../data/gambelii.txt --out ../data/bed/$FILE.m90 --allow-extra-chr
plink --bfile ../data/bed/$FILE.m90 --bp-space 5000 --make-bed --out ../data/bed/$FILE.m90.bp5000 --allow-extra-chr
plink2 --bfile ../data/bed/$FILE.m90 --bp-space 5000 --export phylip --out ../data/bed/$FILE.m90.bp5000 --allow-extra-chr

awk '{$1="0";print $0}' ../data/bed/$FILE.m90.bp5000.bim > ../data/bed/$FILE.m90.bp5000.bim.tmp
mv ../data/bed/$FILE.m90.bp5000.bim.tmp ../data/bed/$FILE.m90.bp5000.bim
```
