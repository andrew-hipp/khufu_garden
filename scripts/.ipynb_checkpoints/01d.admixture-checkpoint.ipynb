## Admixture
We'll run admixture now on both bed files we generated, with 10-fold cross-validation to estimate _K_. I'm using 8 cores because this is not a very large dataset. Because Admixture outputs its analysis files to the working directory, we'll hop over to the output directory and run there, then hope back. First the imputed data:

```bash
FILE=KHU075_SNPcalls_20241105.imputed.bp5000

cd ../out/admixture/imputed
# for i in {1..10};
for i in {8..10};
do
    echo ... doing k=$i....
    admixture --cv=10 ../../../data/bed/$FILE.bed ${i} -j30 > log.$FILE.${i}.out
done

echo DONE WITH $FILE. Moving back to scripts folder.

cd ../../../scripts
```

... and then the imputed data with individuals included only if they have at least 90% of data

```bash
FILE=KHU075_SNPcalls_20241105.imputed.m90.bp5000

cd ../out/admixture/imputed.m90
for i in {1..10};
do
    echo ... doing k=$i....
    admixture --cv=10 ../../../data/bed/$FILE.bed ${i} -j30 > log.$FILE.${i}.out
done

echo DONE WITH $FILE. Moving back to scripts folder.

cd ../../../scripts
```

Then the raw data:

```bash
FILE=KHU075_SNPcalls_20241105.raw.m90.bp5000

# cd ../out/admixture/raw.m90
for i in {1..10};
do
    echo ... doing k=$i....
    admixture --cv=10 ../../../data/bed/$FILE.bed ${i} -j30 > log.$FILE.${i}.out
done

echo DONE WITH $FILE. Moving back to scripts folder.

cd ../../../scripts
```

Then we check the CV error on each of our sets of runs:

```bash
# awk '/CV/ {print $3,$4}' ../out/admixture/raw/*out | cut -c 1-6,7-20 > ../out/admixture/raw/raw.cv.txt
awk '/CV/ {print $3,$4}' ../out/admixture/raw.m90/*out | cut -c 1-6,7-20 > ../out/admixture/raw.m90/raw.m90.cv.txt
awk '/CV/ {print $3,$4}' ../out/admixture/imputed/*out | cut -c 1-6,7-20 > ../out/admixture/imputed/imputed.cv.txt
awk '/CV/ {print $3,$4}' ../out/admixture/imputed.m90/*out | cut -c 1-6,7-20 > ../out/admixture/imputed.m90/imputed.m90.cv.txt


# echo Raw CV
# cat ../out/admixture/raw/raw.cv.txt
# echo

echo Raw m90 CV
cat ../out/admixture/raw.m90/raw.m90.cv.txt
echo

echo
echo Imputed CV
cat ../out/admixture/imputed/imputed.cv.txt
echo

echo
echo Imputed m90 CV
cat ../out/admixture/imputed.m90/imputed.m90.cv.txt
```
