## Data cleanup

### Converting data to vcf
Data were generated using shallow-coverage genome skimming (&half;&times; &ndash; 1&times; average coverage); mapped to a published _Quercus mongolica_ genome (CITATION), and exported in HapMap. Two files were generated: one with just raw data, and one with imputed data as well. Both were converted to vcf files using Tassel and saved in `data/KHUFU`.

### Basic vcf stats
We'll generate stats now using `bcftools +smpl-stats`, but we'll read them into R later to summarize.

````bash
bcftools +smpl-stats ../data/KHUFU/KHU075_SNPcalls2_imputed.vcf -o ../out/stats.imputed_vcf.txt
bcftools +smpl-stats ../data/KHUFU/KHU075_SNPcalls2_raw.vcf -o ../out/stats.raw_vcf.txt
````

### Convert vcf files to bed and phylip
Both vcf files were converted to bed files using `plink`. Both are thinned to one snp per 1000 bp, following the observation that LD tails off at 500 bp in _Quercus_ (Plomion et al., 2016 -- AH check this reference). We also convert to relaxed Phylip format to get genetic distances using RAxML. We are going to filter these in a few ways:

* Exclude two samples of _Quercus gambelii_ from both files, because those were included only as an outgroup for tree-rooting. The gambelii individuals are designated in `data/gambelii.txt`.
* 

First, the imputed data:

````bash
FILE=KHU075_SNPcalls2_imputed

plink --vcf ../data/KHUFU/$FILE.vcf --make-bed --remove ../data/gambelii.txt --out ../data/bed/$FILE --allow-extra-chr
plink --bfile ../data/bed/$FILE --bp-space 1000 --make-bed --out ../data/bed/$FILE.bp1000 --allow-extra-chr
plink2 --bfile ../data/bed/$FILE --bp-space 1000 --export phylip --out ../data/bed/$FILE.bp1000 --allow-extra-chr

awk '{$1="0";print $0}' ../data/bed/$FILE.bp1000.bim > ../data/bed/$FILE.bp1000.bim.tmp
mv ../data/bed/$FILE.bp1000.bim.tmp ../data/bed/$FILE.bp1000.bim
````

and then the raw data. For this, we also eliminate first columns then individuals that are missing more than 80% of data.

````bash
FILE=KHU075_SNPcalls2_raw

plink --vcf ../data/KHUFU/$FILE.vcf --make-bed --geno 0.8 --mind 0.8 --remove ../data/gambelii.txt --out ../data/bed/$FILE --allow-extra-chr
plink --bfile ../data/bed/$FILE --bp-space 1000 --make-bed --out ../data/bed/$FILE.bp1000 --allow-extra-chr
plink2 --bfile ../data/bed/$FILE --bp-space 1000 --export phylip --out ../data/bed/$FILE.bp1000 --allow-extra-chr

awk '{$1="0";print $0}' ../data/bed/$FILE.bp1000.bim > ../data/bed/$FILE.bp1000.bim.tmp
mv ../data/bed/$FILE.bp1000.bim.tmp ../data/bed/$FILE.bp1000.bim
````

###### Admixture
We'll run admixture now on both bed files we generated, with 10-fold cross-validation to estimate _K_. I'm using 8 cores because this is not a very large dataset. Because Admixture outputs its analysis files to the working directory, we'll hop over to the output directory and run there, then hope back. First the imputed data:

````bash
FILE=KHU075_SNPcalls2_imputed.bp1000

cd ../out/admixture/imputed
for i in {1..10};
do
    admixture --cv=10 ../../../data/bed/$FILE.bed ${i} -j8 > log.$FILE.${i}.out
done

cd ../../../scripts
````

Then the raw data:

````bash
FILE=KHU075_SNPcalls2_raw.bp1000

cd ../out/admixture/raw
for i in {1..10};
do
    admixture --cv=10 ../../../data/bed/$FILE.bed ${i} -j8 > log.$FILE.${i}.out
done

cd ../../../scripts
````

Then we check the CV error on each of our sets of runs:

````bash
awk '/CV/ {print $3,$4}' ../out/admixture/raw/*out | cut -c 1-6,7-20 > ../out/admixture/raw/raw.cv.txt
awk '/CV/ {print $3,$4}' ../out/admixture/imputed/*out | cut -c 1-6,7-20 > ../out/admixture/imputed/imputed.cv.txt

echo Raw CV
cat ../out/admixture/raw/raw.cv.txt

echo
echo Imputed CV
cat ../out/admixture/imputed/imputed.cv.txt
````

## Genetic distances
We'll use RAxML to get genetic distances that we can analyze later in R.

````bash
cd ../out
raxmlHPC-PTHREADS -f x -p 75209 -T 20 -s ../data/bed/KHU075_SNPcalls2_imputed.bp1000.phy -m GTRGAMMA -n KHU075_imputed_bp1000
raxmlHPC-PTHREADS -f x -p 75209 -T 20 -s ../data/bed/KHU075_SNPcalls2_raw.bp1000.phy -m GTRGAMMA -n KHU075_raw_bp1000
cd ../scripts
````
